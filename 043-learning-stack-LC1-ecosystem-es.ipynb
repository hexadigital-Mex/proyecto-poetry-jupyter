# Stack de Aprendizaje para Ejercicios con el nuevo Ecosistema LangChain 1.0

## Asegúrate de tener Python 3.13.11 instalado en tu ordenador
* **Consulta las "Lecciones para tu Segunda Ronda" para aprender cómo hacerlo**.

#### Para Usuarios de macOS (esta es la opción que seleccionamos, usa la mejor para ti)
* Actualiza Homebrew y pyenv. Abre tu Terminal y ejecuta:

`brew update && brew upgrade pyenv`

* Busca Python 3.13.11

`pyenv install --list | grep 3.13`

* Instala Python 3.13.11

`pyenv install 3.13.11`

¡Eso es todo! Ya puedes usar esta versión de Python para tus proyectos.

## Asegúrate de tener Poetry 2.2.1 instalado en tu ordenador
* Es posible que tengas Poetry 1.8.2 del stack de aprendizaje inicial, pero quieras probar una versión más nueva (como la 2.2.1). ¡Con pipx, puedes instalar y usar múltiples versiones de Poetry de forma segura al mismo tiempo!

* **Consulta las "Lecciones para tu Segunda Ronda" para aprender cómo hacerlo**.

### Instala la nueva versión de Poetry junto a la antigua, como un comando separado:
`pipx install poetry==2.2.1 --suffix=-2.2.1`

* Esto crea un nuevo comando: `poetry-2.2.1`


#### Asegúrate de tener ambas versiones disponibles
`poetry --version`         # (debería mostrar 1.8.2 o tu versión antigua)
`poetry-2.2.1 --version`  # (debería mostrar 2.2.1)

#### Usa la Versión Correcta para Cada Proyecto
Para los proyectos que requieran Poetry 1.8.2, simplemente sigue usando el comando `poetry` habitual (tu versión antigua).

Para los proyectos del nuevo Ecosistema LangChain 1.0, usa `poetry-2.2.1`.

## Ya estás preparado para trabajar con los archivos de ejercicios del nuevo Ecosistema LangChain 1.0
Seguirás el mismo proceso que aprendiste en las lecciones anteriores para los ejercicios habituales:
* Paso 1: Descarga o clona el Repositorio de Github.
* Paso 2: Navega al Directorio del Proyecto.
* Paso 3: Asegúrate de tener la versión correcta de Python activa en el directorio.
    * `pyenv local 3.13.11` 
* Paso 4: Instala las Dependencias con Poetry.
    * `poetry-2.2.1 install`
* Paso 5: Activa el Entorno Virtual con Poetry (es decir, el Poetry Shell).
    * Recuerda que a partir de Poetry 2.x, la forma de activar el entorno virtual de tu proyecto ha cambiado. Ahora tenemos que hacer:
    * `poetry-2.2.1 env activate`
    * Cuando ejecutes este comando dentro del directorio de tu proyecto, Poetry mostrará el comando que debes usar para activar el entorno en tu terminal actual. Por ejemplo, puede mostrar algo como:
        * `source /Users/tuusuario/Library/Caches/pypoetry/virtualenvs/proyectoincreible-XXXX-py3.13/bin/activate`
    * Copia y ejecuta ese comando en tu terminal.
    * Ahora, tu sesión de terminal está dentro del entorno virtual del proyecto: cualquier comando de Python que ejecutes, o cualquier paquete que instales (por ejemplo, con `poetry-2.2.1 add`), solo afectará a este proyecto—no a tu instalación global de Python.

* Paso 6: Abre los Notebooks de Jupyter incluidos en el proyecto.
    * `jupyter lab` 
* Paso 7: Ejecuta los archivos de Python incluidos en el editor de código.
    * `python nombre_archivo.py`
    * NOTA: Recuerda que debes crear el archivo .env con tus credenciales confidenicales. Puedes utilizar el archivo .env.example como referencia.
* Paso 8: Salir del Poetry Shell.
    * `exit` 
* Paso 9: Reactivar el Poetry Shell.
    * `poetry-2.2.1 env activate`
